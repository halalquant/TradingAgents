# Test Place Order - Sample Request

```python
from tradingagents.dataflows.bybit import place_order
```

```python
# Sample proposal data
# Based on Bybit rules: minOrderAmt is 5 USDT
# 0.01 BTC at ~$100,000 = $1,000 USDT (well above minimum)
proposal = {
    "type": "place_order",
    "id": "2df4d8b0-6a6e-49c6-8c0d-591e3b00ca4a",
    "category": "spot",
    "symbol": "BTCUSDT",
    "side": "Buy",
    "order_type": "Market",
    "qty": 0.01,  # 0.01 BTC = ~$1,000 USD at $100k BTC price
    "market_unit": "baseCoin",
    "price": None,
    "take_profit": None,
    "stop_loss": None
}

print(f"Attempting to place order for {proposal['qty']} BTC...")
print(f"Estimated value at $100k BTC: ${proposal['qty'] * 100000:.2f} USDT")
print(f"Bybit minimum order value: $5 USDT\n")

# Call place_order with the proposal parameters
try:
    result = place_order(
        symbol=proposal["symbol"],
        side=proposal["side"],
        order_type=proposal["order_type"],
        qty=proposal["qty"],
        price=proposal["price"],
        market_unit=proposal["market_unit"],  # 🔧 CRITICAL: Tells Bybit qty is in BTC, not USDT
        stop_loss=proposal["stop_loss"],
        take_profit=proposal["take_profit"],
        category=proposal["category"]
    )
    
    print("✅ Order placed successfully!")
    print("\nOrder Result:")
    import json
    print(json.dumps(result, indent=2))
except ValueError as e:
    print(f"❌ Error: {e}")
    print(f"\nDebug Info:")
    print(f"  Quantity: {proposal['qty']} BTC")
    print(f"  Min Qty: 0.000001 BTC")
    print(f"  Min Order Value: 5 USDT")
    print(f"  Market Unit: {proposal['market_unit']}")
    print(f"\nThis should work. The error might be:")
    print(f"  1. Insufficient balance in your account")
    print(f"  2. Market conditions or Bybit API restrictions")
    print(f"  3. Check your API key permissions")
```

# Test Limit Order - Sample Request

```python
# Sample limit order proposal
# Limit orders require a price and don't use market_unit
# Set price slightly below current market to test (won't execute immediately)

# Get current BTC price and instrument info for proper precision
from tradingagents.dataflows.bybit import bybit_v5_request, place_order
import math

ticker_data = bybit_v5_request("GET", "/v5/market/tickers", {
    "category": "spot",
    "symbol": "BTCUSDT"
})
current_price = float(ticker_data["result"]["list"][0]["lastPrice"])
print(f"Current BTC Price: ${current_price:,.2f}")

# Get tick size (price precision) from instrument info
instrument_data = bybit_v5_request("GET", "/v5/market/instruments-info", {
    "category": "spot",
    "symbol": "BTCUSDT"
})
tick_size = float(instrument_data["result"]["list"][0]["priceFilter"]["tickSize"])
print(f"Tick Size: ${tick_size}")

# Calculate limit price 1% below current, rounded to tick size
target_price = current_price * 0.99
# Round down to nearest tick size multiple
limit_price = math.floor(target_price / tick_size) * tick_size
print(f"Target Price (1% below): ${target_price:,.2f}")
print(f"Limit Price (rounded to tick): ${limit_price:,.2f}")

limit_proposal = {
    "type": "place_order",
    "id": "limit-test-001",
    "category": "spot",
    "symbol": "BTCUSDT",
    "side": "Sell",
    "order_type": "Market",
    "qty": 0.030476,  # 0.001 BTC = ~$100 USD at $100k BTC price
    "price": None,  # Required for limit orders
    "take_profit": None,
    "stop_loss": None,
    "time_in_force": "GTC"  # Good Till Cancelled
}

print(f"\nAttempting to place LIMIT order:")
print(f"  Quantity: {limit_proposal['qty']} BTC")
# print(f"  Limit Price: ${limit_proposal['price']:,.2f}")
# print(f"  Order Value: ${limit_proposal['qty'] * limit_proposal['price']:.2f} USDT")
print(f"  Time in Force: {limit_proposal['time_in_force']}")
print(f"  (Order will execute if BTC drops to ${limit_price:,.2f})")

# Call place_order with limit order parameters
try:
    result = place_order(
        symbol=limit_proposal["symbol"],
        side=limit_proposal["side"],
        order_type=limit_proposal["order_type"],
        qty=limit_proposal["qty"],
        price=limit_proposal["price"],  # Required for Limit orders
        stop_loss=limit_proposal["stop_loss"],
        take_profit=limit_proposal["take_profit"],
        time_in_force=limit_proposal["time_in_force"],
        category=limit_proposal["category"]
        # NOTE: No market_unit for limit orders
    )
    
    print("\n✅ Limit order placed successfully!")
    print("\nOrder Result:")
    import json
    print(json.dumps(result, indent=2))
    
    if "orderId" in result:
        print(f"\n📝 Order ID: {result['orderId']}")
        print(f"💡 You can cancel this order using: cancel_order('{result['orderId']}', 'BTCUSDT')")
    
except ValueError as e:
    print(f"\n❌ Error: {e}")
    print(f"\nDebug Info:")
    print(f"  Quantity: {limit_proposal['qty']} BTC")
    print(f"  Price: ${limit_proposal['price']:,.2f}")
    print(f"  Order Value: ${limit_proposal['qty'] * limit_proposal['price']:.2f} USDT")
    print(f"  Min Order Value: 5 USDT")
    print(f"  Tick Size: ${tick_size}")
    print(f"\nPossible issues:")
    print(f"  1. Insufficient balance")
    print(f"  2. Price not within allowed range")
    print(f"  3. API key permissions")
```

# Test Amend Order

```python
# Amend (modify) an existing limit order
# You can change qty, price, stop_loss, or take_profit

from tradingagents.dataflows.bybit import amend_order

# If you ran the limit order cell, the order_id should be available
try:
    if 'result' in dir() and isinstance(result, dict) and 'orderId' in result:
        order_id = result['orderId']
        
        # Get current price to set a new limit price
        ticker_data = bybit_v5_request("GET", "/v5/market/tickers", {
            "category": "spot",
            "symbol": "BTCUSDT"
        })
        current_price = float(ticker_data["result"]["list"][0]["lastPrice"])
        
        # Set new limit price 2% below current (more aggressive than before)
        new_limit_price = round(current_price * 0.98, 2)
        new_qty = 0.0015  # Increase quantity from 0.001 to 0.0015
        
        print(f"Amending order: {order_id}")
        print(f"  New Quantity: {new_qty} BTC (was 0.001 BTC)")
        print(f"  New Price: ${new_limit_price:,.2f} (2% below market)")
        print(f"  New Value: ${new_qty * new_limit_price:.2f} USDT\n")
        
        amend_result = amend_order(
            order_id=order_id,
            symbol="BTCUSDT",
            qty=new_qty,  # Update quantity
            price=new_limit_price,  # Update price
            category="spot"
        )
        
        print("✅ Order amended successfully!")
        print("\nAmend Result:")
        import json
        print(json.dumps(amend_result, indent=2))
        
    else:
        print("⚠️  No order_id found.")
        print("Please run the limit order cell first to create an order, then amend it.")
        print("\nManual example:")
        print("  amend_result = amend_order(")
        print("      order_id='YOUR_ORDER_ID',")
        print("      symbol='BTCUSDT',")
        print("      qty=0.002,  # New quantity")
        print("      price=95000  # New price")
        print("  )")
        
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nNote: Only unfilled or partially filled orders can be amended.")
    print("If the order already executed, you'll need to place a new order.")
```

# Check Bybit Instrument Info for BTCUSDT

```python
from tradingagents.dataflows.bybit import bybit_v5_request
import json

# Check instrument info for BTCUSDT spot
params = {
    "category": "spot",
    "symbol": "BTCUSDT"
}

data = bybit_v5_request("GET", "/v5/market/instruments-info", params)
instrument = data.get("result", {}).get("list", [])

if instrument:
    info = instrument[0]
    print("BTCUSDT Spot Trading Rules:")
    print(f"Symbol: {info.get('symbol')}")
    print(f"Base Coin: {info.get('baseCoin')}")
    print(f"Quote Coin: {info.get('quoteCoin')}")
    print(f"\nLot Size (Quantity) Rules:")
    lot_size = info.get('lotSizeFilter', {})
    print(f"  Min Order Qty: {lot_size.get('minOrderQty')} BTC")
    print(f"  Max Order Qty: {lot_size.get('maxOrderQty')} BTC")
    print(f"  Qty Step: {lot_size.get('basePrecision')}")
    
    # Check for minimum notional value
    print(f"\nMinimum Order Value:")
    if 'minNotionalValue' in lot_size:
        print(f"  Min Notional: {lot_size.get('minNotionalValue')} USDT")
    else:
        print(f"  No minNotionalValue found in lotSizeFilter")
    
    print(f"\nPrice Rules:")
    price_filter = info.get('priceFilter', {})
    print(f"  Min Price: {price_filter.get('minPrice')}")
    print(f"  Max Price: {price_filter.get('maxPrice')}")
    print(f"  Tick Size: {price_filter.get('tickSize')}")
    
    print("\n\nFull Instrument Info:")
    print(json.dumps(info, indent=2))
else:
    print("Could not retrieve instrument info")
```
